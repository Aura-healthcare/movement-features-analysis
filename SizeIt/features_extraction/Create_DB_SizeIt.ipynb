```python
def get_all_tsv(dir_patient_eeg):
    
    l_files = []
    dossier = Path(dir_patient_eeg)
    tsv_files = list(dossier.glob("*.tsv"))
    
    for f in tsv_files:
        l_files.append(f)

    return l_files

def get_patient_number(adresse_eeg):
    pos = adresse_eeg.find("sub")
    result = adresse_eeg[pos + 4 : pos + 7]
    return result

dir_all_pat = '/data2/datasets/seizeit2-dataset/ds005873-1.1.0/'
l_patients = os.listdir(dir_all_pat) 
l_dir_patients = [dir_all_pat + '/' + patient + '/ses-01/' for patient in l_patients]
l_dir_patients_eeg = [toto + 'eeg/' for toto in l_dir_patients]
# I only keep directory with patients 
l_dir_patients_eeg = [element for element in l_dir_patients_eeg if "sub" in element]
```

# créer un dataframe avec : 
# sz_type | no patient | run | adresse_file | deb_sec | durée | 

```python
df_events = pd.DataFrame(columns=['onset', 'duration', 'eventType', 'lateralization', 'localization', 'vigilance', 'confidence', 
'channels', 'dateTime', 'recordingDuration','adresse_ecg','adresse_acc','patient'])


for eeg in l_dir_patients_eeg:
    l_tsv = get_all_tsv(eeg)
    
    for path_tsv in l_tsv:
        path_ecg = str(path_tsv)[:-10] + 'ecg.edf'
        path_mov = str(path_tsv)[:-10] + 'mov.edf'
        path_ecg = path_ecg.replace("eeg","ecg")
        path_mov = path_mov.replace("eeg","mov")
        patient = get_patient_number(eeg)

        with open(path_tsv, "r", encoding="utf-8") as f:
            for i,ligne in enumerate(f):
                ligne = ligne[0:-1]
                if i > 0:
                    df_events.loc[len(df_events)] = ligne.split("\t") + [path_ecg] + [path_mov] + [patient]
        
df_events.to_csv('Datas/bdd_SizeIt.csv')
```
