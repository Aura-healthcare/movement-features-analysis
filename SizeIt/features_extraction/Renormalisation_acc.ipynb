```python
# Fonctions

def renormalisation_norm(df,cols_acc,norme):
    
    df[norme] = np.linalg.norm(
    df[cols_acc].values,
    axis=1
    )
    renorm_factor = df[norme].mean()
    df[cols_acc] *= (1/ renorm_factor)
    df[norme] = np.linalg.norm(
    df[cols_acc].values,
    axis=1
    )

    return df, renorm_factor
```

```python

input_folder = '/home/aura-fabien/SizeIT/Datas/Csv_acc/'
output_folder = '/home/aura-fabien/SizeIT/Datas/Csv_acc_renormalised/'
log_path = "/home/aura-fabien/SizeIT/Datas/Statistiques/renormalisation.log" 

bdd_seizure_modele = pd.read_csv('/home/aura-fabien/SizeIT/Datas/Preprocessing_datas/bdd_seizure_modele.csv')
l_csv_files = os.listdir(input_folder)
l_patient_modele = bdd_seizure_modele.patient.unique()

l_acc_1 = ['EEG SD ACC X','EEG SD ACC Y', 'EEG SD ACC Z']
l_acc_2 = ['ECGEMG SD ACC X','ECGEMG SD ACC Y','ECGEMG SD ACC Z']

df_stats_renormalisation = pd.DataFrame(columns = ['factor_renor_1','factor_renor_2'])
```

```python
deb = time.time()

# --- Configuration du log ---


logging.basicConfig(
    filename=log_path,
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s"
)

logging.info("Début de la procédure de renormalisation")

for no_file,file in enumerate(l_csv_files):

    if no_file % 100 == 0:
        print('no file = ', no_file)

    try:
        df_acc = pd.read_csv(input_folder + file) 
        df_acc, renorm_1 = renormalisation_norm(df_acc,l_acc_1,'norme_1')
        df_acc, renorm_2 = renormalisation_norm(df_acc,l_acc_2,'norme_2')
        df_stats_renormalisation.loc[no_file,['factor_renor_1','factor_renor_2']] = [renorm_1,renorm_2]
        df_acc.to_csv(output_folder + file )
        df_stats_renormalisation.to_csv('/home/aura-fabien/SizeIT/Datas/Statistiques/' + 'renormalisation.csv')
    
    except Exception as e:
        logging.error(f"Erreur avec le fichier {file}")
        logging.exception(e) 
        print(df_acc.head(1))

print('temps ecoulé = ', time.time() - deb)
```
